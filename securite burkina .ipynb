```python
# 1. Chargement du Dataset réel (Janvier à Août 2026)
data = {
    'Mois': ['Janvier', 'Février', 'Mars', 'Avril', 'Mai', 'Juin', 'Juillet', 'Août'],
    'Accidents_Reels': [1945, 1794, 1803, 1992, 1783, 1826, 1744, 1516],
    'Blesses_Reels': [1728, 1337, 1310, 1533, 1552, 1537, 1459, 1347],
    'Deces_Reels': [123, 83, 99, 105, 148, 91, 113, 79]
}
df = pd.DataFrame(data)

# 2. Simulation de la tendance de fin d'année (Septembre à Décembre)
# On calcule les moyennes historiques pour projeter les mois manquants
moy_acc = int(df['Accidents_Reels'].mean())
moy_ble = int(df['Blesses_Reels'].mean())
moy_dec = int(df['Deces_Reels'].mean())

# Ajout de facteurs contextuels de base (ex: hausse en Octobre pour la rentrée, forte hausse en Décembre pour les fêtes)
fin_annee = pd.DataFrame({
    'Mois': ['Septembre', 'Octobre', 'Novembre', 'Décembre'],
    'Accidents_Reels': [moy_acc, moy_acc + 100, moy_acc, moy_acc + 250],
    'Blesses_Reels': [moy_ble, moy_ble + 80, moy_ble, moy_ble + 200],
    'Deces_Reels': [moy_dec, moy_dec + 10, moy_dec, moy_dec + 35]
})
df_complet = pd.concat([df, fin_annee], ignore_index=True)

# 3. Définition des paramètres des nouvelles politiques publiques
REDUCTION_CAMERA = 0.20  # Les caméras réduisent le volume des accidents de 20%
REDUCTION_CASQUE = 0.30  # Le port obligatoire du casque réduit les décès de 30%

# 4. Calcul des prédictions (Scénario avec politiques appliquées)
df_complet['Accidents_Predits'] = df_complet['Accidents_Reels'] * (1 - REDUCTION_CAMERA)
df_complet['Deces_Predits'] = df_complet['Deces_Reels'] * (1 - REDUCTION_CASQUE)

# 5. Ajout d'une variable d'ajustement : Taux d'application réel de la loi
# Si le décret n'est respecté qu'à 80%, on ajuste la courbe de prédiction
taux_respect = 0.80  
df_complet['Deces_Ajustes_Reels'] = df_complet['Deces_Reels'] - (df_complet['Deces_Reels'] - df_complet['Deces_Predits']) * taux_respect

# Arrondir les résultats pour obtenir des nombres entiers de personnes/accidents
df_complet[['Accidents_Predits', 'Deces_Predits', 'Deces_Ajustes_Reels']] = df_complet[['Accidents_Predits', 'Deces_Predits', 'Deces_Ajustes_Reels']].round(0).astype(int)

# Affichage des résultats comparatifs
print("=== SIMULATION IMPACT DES POLITIQUES ROUTIÈRES (BURKINA FASO 2026) ===")
print(df_complet[['Mois', 'Accidents_Reels', 'Accidents_Predits', 'Deces_Reels', 'Deces_Predits', 'Deces_Ajustes_Reels']])

# Calcul de l'impact global à l'échelle de l'année
vies_sauvees = df_complet['Deces_Reels'].sum() - df_complet['Deces_Ajustes_Reels'].sum()
print(f"\n[RÉSULTAT PROJECTION] : Cette simulation prévoit environ {vies_sauvees} vies sauvées sur l'année grâce au port du casque.")
```

```python
# 1. Chargement du Dataset réel (Janvier à Août 2026)
data = {
    'Mois': ['Janvier', 'Février', 'Mars', 'Avril', 'Mai', 'Juin', 'Juillet', 'Août'],
    'Accidents_Reels': [1945, 1794, 1803, 1992, 1783, 1826, 1744, 1516],
    'Blesses_Reels': [1728, 1337, 1310, 1533, 1552, 1537, 1459, 1347],
    'Deces_Reels': [123, 83, 99, 105, 148, 91, 113, 79]
}
df = pd.DataFrame(data)

# 2. Paramètres des Facteurs Exogènes
CROISSANCE_ANNUELLE = 0.04  # +4% de croissance démographique et du parc auto par an
taux_mensuel_croissance = CROISSANCE_ANNUELLE / 12

REDUCTION_CAMERA = 0.20     # Les caméras réduisent les accidents de 20%
REDUCTION_CASQUE = 0.30     # Le casque réduit le risque de décès de 30%
TAUX_RESPECT_LOI = 0.80     # 80% de la population respecte les mesures

# 3. Projection de la fin d'année (Septembre à Décembre) avec Facteur Démographique et Saisonnalité
moy_acc = df['Accidents_Reels'].mean()
moy_ble = df['Blesses_Reels'].mean()
moy_dec = df['Deces_Reels'].mean()

mois_futurs = ['Septembre', 'Octobre', 'Novembre', 'Décembre']
fin_annee_list = []

for i, mois in enumerate(mois_futurs, start=9): # start=9 car Septembre est le 9ème mois
    # Calcul de l'effet cumulé de la croissance démographique au fil des mois
    facteur_demographique = 1 + (taux_mensuel_croissance * i)
    
    # Intégration de la saisonnalité (ex: hausse des fêtes en Décembre)
    facteur_saison_acc = 1.15 if mois == 'Décembre' else (1.05 if mois == 'Octobre' else 1.0)
    facteur_saison_dec = 1.25 if mois == 'Décembre' else 1.0
    
    base_acc = moy_acc * facteur_demographique * facteur_saison_acc
    base_ble = moy_ble * facteur_demographique * facteur_saison_acc
    base_dec = moy_dec * facteur_demographique * facteur_saison_dec
    
    fin_annee_list.append({
        'Mois': mois,
        'Accidents_Reels': int(base_acc),
        'Blesses_Reels': int(base_ble),
        'Deces_Reels': int(base_dec)
    })

df_complet = pd.concat([df, pd.DataFrame(fin_annee_list)], ignore_index=True)

# 4. Application des Politiques Publiques (Scénario Prédictif)
# L'effet démographique est inclus dans "Accidents_Reels" car la population augmente la base de crashs
df_complet['Accidents_Predits'] = df_complet['Accidents_Reels'] * (1 - (REDUCTION_CAMERA * TAUX_RESPECT_LOI))

# Pour les décès, on calcule d'abord la prédiction théorique avec casque
deces_theorique_casque = df_complet['Deces_Reels'] * (1 - REDUCTION_CASQUE)
# Puis on applique le taux de respect réel sur le terrain
df_complet['Deces_Predits'] = df_complet['Deces_Reels'] - (df_complet['Deces_Reels'] - deces_theorique_casque) * TAUX_RESPECT_LOI

# Arrondir pour les humains
df_complet[['Accidents_Predits', 'Deces_Predits']] = df_complet[['Accidents_Predits', 'Deces_Predits']].round(0).astype(int)

# 5. Résultat
vies_sauvees = df_complet['Deces_Reels'].sum() - df_complet['Deces_Predits'].sum()
accidents_evites = df_complet['Accidents_Reels'].sum() - df_complet['Accidents_Predits'].sum()

print("=== PROJECTION 2026 AVEC FACTEUR DÉMOGRAPHIQUE (+4%/an) ===")
print(df_complet[['Mois', 'Accidents_Reels', 'Accidents_Predits', 'Deces_Reels', 'Deces_Predits']])
print(f"\n[BILAN ANNUEL AJUSTÉ] :")
print(f"-> Accidents évités grâce aux caméras (ajusté démo) : {accidents_evites} crashs")
print(f"-> Vies sauvées grâce au casque (ajusté démo) : {vies_sauvees} personnes")
```

```python
# 1. Loading Real Dataset (January to August 2026)
data = {
    'Mois': ['Janvier', 'Février', 'Mars', 'Avril', 'Mai', 'Juin', 'Juillet', 'Août'],
    'Accidents_Reels': [1945, 1794, 1803, 1992, 1783, 1826, 1744, 1516],
    'Blesses_Reels': [1728, 1337, 1310, 1533, 1552, 1537, 1459, 1347],
    'Deces_Reels': [123, 83, 99, 105, 148, 91, 113, 79]
}
df = pd.DataFrame(data)

# 2. Vector Parameters & Structural Assumptions
CROISSANCE_ANNUELLE = 0.04             # Vector 1: Demographic growth & vehicle influx (+4%/year)
taux_mensuel_croissance = CROISSANCE_ANNUELLE / 12

REDUCTION_CAMERA = 0.20                # Vector 2: Technology - Camera raw crash volume reduction
REDUCTION_CASQUE = 0.30                # Vector 3: Behavioral - Helmet protection raw mortality reduction
TAUX_RESPECT_LOI = 0.80                # Variable: Public law compliance rate (20% incivility buffer)

IMPACT_MAUVAISES_INTERSECTIONS = 0.12  # Vector 4: Infrastructure - Blind spots & faulty road design (+12% volume)
DEGRADATION_ROUTES_PLUIE = 0.15        # Vector 5: Environment - Potholes & heavy rains during rainy season (+15% severity)

# 3. Monthly Progression Loop (Applying all parameters systematically)
df_resultats = []
moy_acc = df['Accidents_Reels'].mean()
moy_ble = df['Blesses_Reels'].mean()
moy_dec = df['Deces_Reels'].mean()

# Complete list of 12 months for 2026 simulation
tous_les_mois = [
    ('Janvier', True), ('Février', True), ('Mars', True), ('Avril', True),
    ('Mai', True), ('Juin', True), ('Juillet', True), ('Août', True),
    ('Septembre', False), ('Octobre', False), ('Novembre', False), ('Décembre', False)
]

for index_mois, (mois, is_real) in enumerate(tous_les_mois, start=1):
    # Base calculation (Real historical value or historical average prediction)
    if is_real:
        acc_base = df.loc[df['Mois'] == mois, 'Accidents_Reels'].values[0]
        ble_base = df.loc[df['Mois'] == mois, 'Blesses_Reels'].values[0]
        dec_base = df.loc[df['Mois'] == mois, 'Deces_Reels'].values[0]
    else:
        # Holiday seasonality weights for the projected end of the year
        facteur_saison_acc = 1.15 if mois == 'Décembre' else (1.05 if mois == 'Octobre' else 1.0)
        facteur_saison_dec = 1.25 if mois == 'Décembre' else 1.0
        
        acc_base = moy_acc * facteur_saison_acc
        ble_base = moy_ble * facteur_saison_acc
        dec_base = moy_dec * facteur_saison_dec

    # Vector 1: Compounding Demographic Growth over time
    facteur_demographique = 1 + (taux_mensuel_croissance * index_mois)
    acc_base_ajustee = acc_base * facteur_demographique
    dec_base_ajustee = dec_base * facteur_demographique
    
    # Vector 4 & 5: Environmental & Infrastructure Penalties
    facteur_intersections = 1 + IMPACT_MAUVAISES_INTERSECTIONS
    facteur_pluie_route = 1 + DEGRADATION_ROUTES_PLUIE if mois in ['Juillet', 'Août', 'Septembre'] else 1.0

    # Vector 2 & 3: Final Policy Execution Formula (Law vs Environment)
    acc_predit = acc_base_ajustee * (1 - (REDUCTION_CAMERA * TAUX_RESPECT_LOI)) * facteur_intersections
    dec_predit = dec_base_ajustee * (1 - (REDUCTION_CASQUE * TAUX_RESPECT_LOI)) * facteur_pluie_route
    
    df_resultats.append({
        'Mois': mois,
        'Accidents_Sans_Mesures': int(round(acc_base_ajustee)),
        'Accidents_Predits_Pol': int(round(acc_predit)),
        'Deces_Sans_Mesures': int(round(dec_base_ajustee)),
        'Deces_Predits_Pol': int(round(dec_predit))
    })

df_final = pd.DataFrame(df_resultats)

# 4. Global Simulation Summary Printout
print("==========================================================================")
print("     SIMULATION DES SÉCURITÉS ROUTIÈRES COMPLÈTE - BURKINA FASO 2026")
print("==========================================================================")
print(df_final.to_string(index=False))
print("==========================================================================")

total_acc_sans = df_final['Accidents_Sans_Mesures'].sum()
total_acc_avec = df_final['Accidents_Predits_Pol'].sum()
total_dec_sans = df_final['Deces_Sans_Mesures'].sum()
total_dec_avec = df_final['Deces_Predits_Pol'].sum()

print(f"-> Accidents évitables malgré les carrefours défectueux : {total_acc_sans - total_acc_avec} crashs")
print(f"-> Vies humaines sauvées malgré les routes dégradées     : {total_dec_sans - total_dec_avec} décès évités")
print("==========================================================================")
```

```python
# Set random seed for exact reproducibility of the random baseline model
np.random.seed(42)

# ==========================================
# 1. LOAD HISTORICAL DATASET (Jan - Aug 2026)
# ==========================================
data = {
    'Mois': ['Janvier', 'Février', 'Mars', 'Avril', 'Mai', 'Juin', 'Juillet', 'Août'],
    'Accidents_Reels': [1945, 1794, 1803, 1992, 1783, 1826, 1744, 1516],
    'Blesses_Reels': [1728, 1337, 1310, 1533, 1552, 1537, 1459, 1347],
    'Deces_Reels': [123, 83, 99, 105, 148, 91, 113, 79]
}
df_historical = pd.DataFrame(data)

# Calculate historical averages to project baseline for unobserved months (Q4)
avg_acc = df_historical['Accidents_Reels'].mean()
avg_dec = df_historical['Deces_Reels'].mean()

# ==========================================
# 2. MODEL PARAMETERS & BEHAVIORAL VECTORS
# ==========================================
# Vector 1: Compounding Demographic/Vehicle growth (+4% per year)
ANNUAL_GROWTH = 0.04             
monthly_growth_rate = ANNUAL_GROWTH / 12

# Vector 2 & 3: Legislative Intentions & Behavioral Changes
CAMERA_REDUCTION = 0.20          # Speed cameras reduce crude crash volume by 20%
HELMET_REDUCTION = 0.30          # Mandatory helmets reduce crude death rate by 30%
LAW_COMPLIANCE = 0.80            # 80% public compliance rate (20% incivility buffer)

# Vector 4 & 5: Infrastructure & Environmental Friction Elements
FAULTY_INTERSECTIONS = 0.12      # Flawed layout design adds +12% incompressible crash volume
RAINY_SEASON_POTHOLES = 0.15     # Rainy season (Hivernage) degradation adds +15% mortality risk

# Random Baseline Margin of Variance Bound (Calculated from Historical Average Abs |Δ%|)
RANDOM_BOUND = 0.08              

# ==========================================
# 3. COMPLETE 12-MONTH ITERATIVE CORE LOOP
# ==========================================
full_year_months = [
    ('Janvier', True), ('Février', True), ('Mars', True), ('Avril', True),
    ('Mai', True), ('Juin', True), ('Juillet', True), ('Août', True),
    ('Septembre', False), ('Octobre', False), ('Novembre', False), ('Décembre', False)
]

simulation_results = []

# Tracks for Random Walk Simulation
last_acc_random = df_historical['Accidents_Reels'].iloc[-1]
last_dec_random = df_historical['Deces_Reels'].iloc[-1]

for index, (mois, is_historical) in enumerate(full_year_months, start=1):
    
    # --- PHASE A: CHOOSE BASE CHANNELS ---
    if is_historical:
        acc_base = df_historical.loc[df_historical['Mois'] == mois, 'Accidents_Reels'].values[0]
        dec_base = df_historical.loc[df_historical['Mois'] == mois, 'Deces_Reels'].values[0]
    else:
        # Standard Q4 calendar weight filters (School openings in Oct, holidays in Dec)
        season_acc_weight = 1.15 if mois == 'Décembre' else (1.05 if mois == 'Octobre' else 1.0)
        season_dec_weight = 1.25 if mois == 'Décembre' else 1.0
        
        acc_base = avg_acc * season_acc_weight
        dec_base = avg_dec * season_dec_weight

    # --- PHASE B: STRUCTURAL CORRECTION MODEL ---
    # 1. Apply Vector 1 (Demographics over time index)
    demographic_factor = 1 + (monthly_growth_rate * index)
    acc_dem_adjusted = acc_base * demographic_factor
    dec_dem_adjusted = dec_base * demographic_factor
    
    # 2. Identify Infrastructure & Environmental Penalties
    intersection_penalty = 1 + FAULTY_INTERSECTIONS
    pothole_penalty = 1 + RAINY_SEASON_POTHOLES if mois in ['Juillet', 'Août', 'Septembre'] else 1.0
    
    # 3. Apply Multi-Vector Collision Equations
    # Volume reduction from cameras checked by engineering errors
    acc_structural = acc_dem_adjusted * (1 - (CAMERA_REDUCTION * LAW_COMPLIANCE)) * intersection_penalty
    # Mortality reduction from helmets checked by weather-driven asphalt damage
    dec_structural = dec_dem_adjusted * (1 - (HELMET_REDUCTION * LAW_COMPLIANCE)) * pothole_penalty

    # --- PHASE C: RANDOM WALK BENCHMARK MODEL ---
    if index <= 8:
        # Mirrors historical facts as a starting platform canvas
        acc_random = acc_base
        dec_random = dec_base
    else:
        # Mutates completely using the +/- 8% realistic variance rule
        acc_random = last_acc_random * (1 + np.random.uniform(-RANDOM_BOUND, RANDOM_BOUND))
        dec_random = last_dec_random * (1 + np.random.uniform(-RANDOM_BOUND, RANDOM_BOUND))
        
        # Cache current positions for next step propagation
        last_acc_random = acc_random
        last_dec_random = dec_random

    # Append mapped parameters into results matrix
    simulation_results.append({
        'Mois': mois,
        'Accidents_Statu_Quo': int(round(acc_base if is_historical else acc_dem_adjusted)),
        'Accidents_Structurel': int(round(acc_structural)),
        'Accidents_Random': int(round(acc_random)),
        'Deces_Statu_Quo': int(round(dec_base if is_historical else dec_dem_adjusted)),
        'Deces_Structurel': int(round(dec_structural)),
        'Deces_Random': int(round(dec_random))
    })

# Convert list to Pandas DataFrame
df_dashboard = pd.DataFrame(simulation_results)

# ==========================================
# 4. PRINT SUMMARY ANALYSIS REPORT
# ==========================================
print("\n" + "="*90)
print("     BILAN COMPARATIF REEL & SIMULATION : MODÈLE STRUCTUREL VS BASELINE RANDOM")
print("="*90)
print(df_dashboard.to_string(index=False))
print("="*90)

# Calculate annual savings and metrics
saved_accidents = df_dashboard['Accidents_Statu_Quo'].sum() - df_dashboard['Accidents_Structurel'].sum()
saved_vies = df_dashboard['Deces_Statu_Quo'].sum() - df_dashboard['Deces_Structurel'].sum()

print(f" -> Nombre total de crashs évités grâce aux lois (ajusté carrefours) : {saved_accidents} accidents.")
print(f" -> Nombre total de vies humaines sauvées sur l'année (ajusté routes) : {saved_vies} personnes.")
print("="*90 + "\n")
```
